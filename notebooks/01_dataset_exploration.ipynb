```python
legitimate = (df["Class"] == 0).sum()
fraud = (df["Class"] == 1).sum()

print("Total transactions:", len(df))
print("Legitimate transactions:", legitimate)
print("Fraudulent transactions:", fraud)
print("Fraud percentage:", round((fraud / len(df)) * 100, 4), "%")
```

```python
# Initial Dataset Findings

# The dataset contains 284,807 financial transactions.

# - Legitimate transactions: 284,315
# - Fraudulent transactions: 492
# - Fraud rate: 0.1727%

# The target variable `Class` is highly imbalanced, with fraudulent
# transactions representing only a very small proportion of the dataset.

# This class imbalance will be an important consideration during model
# training and evaluation. Therefore, accuracy alone will not be sufficient
# to evaluate our fraud detection models. We will also consider precision,
# recall, F1-score, ROC-AUC and PR-AUC.
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))

sns.boxplot(data=df, x="Class", y="Amount")

plt.title("Transaction Amount Distribution by Class")
plt.xlabel("Class (0 = Legitimate, 1 = Fraud)")
plt.ylabel("Transaction Amount")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(data=df, x="Class", y="Amount")

plt.yscale("log")
plt.title("Transaction Amount Distribution by Class (Log Scale)")
plt.xlabel("Class (0 = Legitimate, 1 = Fraud)")
plt.ylabel("Transaction Amount (log scale)")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

ax = sns.countplot(data=df, x="Class")

plt.title("Transaction Class Distribution")
plt.xlabel("Class (0 = Legitimate, 1 = Fraud)")
plt.ylabel("Number of Transactions")

for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="Time",
    hue="Class",
    bins=100,
    element="step",
    stat="density",
    common_norm=False
)

plt.title("Transaction Time Distribution by Class")
plt.xlabel("Time")
plt.ylabel("Density")

plt.show()
```

```python
df["TimeBin"] = pd.cut(df["Time"], bins=20)

fraud_rate_by_time = df.groupby("TimeBin", observed=True)["Class"].mean()

fraud_rate_by_time
```

```python
plt.figure(figsize=(12, 5))

fraud_rate_by_time.plot(kind="line", marker="o")

plt.title("Fraud Rate Across Transaction Time")
plt.xlabel("Time Bin")
plt.ylabel("Fraud Rate")

plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
# Transaction Time Analysis

# The fraud rate varies across different transaction-time intervals rather than
# remaining constant throughout the dataset.

# Several time intervals show higher fraud rates than others, suggesting that
# transaction timing may contain useful information for fraud detection.

# These results represent associations in the dataset and do not imply that
# transaction time directly causes fraudulent behavior.
```

```python
correlation = df.corr(numeric_only=True)["Class"].sort_values(ascending=False)

correlation
```

```python
plt.figure(figsize=(10, 8))

sns.barplot(
    x=correlation.values,
    y=correlation.index
)

plt.title("Feature Correlation with Fraud Class")
plt.xlabel("Correlation with Class")
plt.ylabel("Feature")

plt.show()
```

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    df.drop(columns=["TimeBin"]).corr(numeric_only=True),
    cmap="coolwarm",
    center=0
)

plt.title("Feature Correlation Heatmap")

plt.show()
```

```python
# Correlation Analysis

# The correlation heatmap shows the relationships among the numerical features
# and the fraud target.

# Several PCA-transformed features show stronger positive or negative
# correlations with `Class`, while many features have correlations close to
# zero.

# The V1–V28 features have relatively low correlation with one another. This is
# consistent with their PCA-transformed representation.

# Because V1–V28 are anonymized PCA components, their individual business
# interpretations are not available. Therefore, correlation values are used
# for statistical analysis rather than assigning business meaning to specific
# components.
```

```python
class_corr = (
    df.drop(columns=["TimeBin"], errors="ignore")
      .corr(numeric_only=True)["Class"]
      .drop("Class")
      .sort_values()
)

top_corr = pd.concat([
    class_corr.head(10),
    class_corr.tail(10)
])

plt.figure(figsize=(10, 8))

sns.barplot(
    x=top_corr.values,
    y=top_corr.index
)

plt.title("Features Most Correlated with Fraud")
plt.xlabel("Correlation with Class")
plt.ylabel("Feature")

plt.show()
```

```python
# Features Associated with Fraud

# The correlation analysis identifies several PCA-transformed features with
# stronger positive or negative linear relationships with the fraud target.

# Features such as V17, V14 and V12 show relatively stronger negative
# correlations, while V11, V4 and V2 show relatively stronger positive
# correlations.

# These correlations describe statistical relationships with the target and
# should not be interpreted as model feature importance. Later stages of
# FraudLens will use model-based feature importance and SHAP to investigate
# which features contribute most to predictions.
```

```python
# EDA Conclusions

# The initial exploration of the Credit Card Fraud Detection dataset produced
# several important findings:

# 1. The dataset contains 284,807 transactions, of which only 492 are
#    fraudulent.

# 2. Fraudulent transactions represent approximately 0.1727% of all
#    transactions, creating a severe class imbalance problem.

# 3. Transaction amounts are strongly right-skewed and contain extreme values.
#    Legitimate transactions have a particularly large range of transaction
#    amounts.

# 4. Transaction amount distributions overlap between legitimate and
#    fraudulent transactions, so amount alone cannot reliably identify fraud.

# 5. Fraud rates vary across transaction-time intervals, suggesting that
#    temporal information may contain predictive signals.

# 6. Several PCA-transformed features show stronger positive or negative
#    correlations with the fraud target.

# 7. The V1–V28 variables are anonymized PCA components, so their individual
#    business meanings cannot be directly interpreted.

# These findings will guide the preprocessing, feature engineering, model
# selection and evaluation stages of FraudLens.
```
